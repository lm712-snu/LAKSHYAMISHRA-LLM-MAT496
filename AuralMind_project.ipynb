# AuralMind 2.0 — Project Notebook
This notebook walks through the scaffold, how to run local prototypes, and contains runnable code where possible.

## Setup
Install dependencies (preferably in a venv):

```
pip install -r requirements.txt
```

Some libraries (librosa, openl3) may need system-level packages.

## Files in this scaffold
- langgraph.yaml
- fastapi_app.py
- streamlit_app.py
- seed_data.py
- utils.py

Open and edit `fastapi_app.py` to integrate real agents and API calls.

## Running the FastAPI skeleton
In a terminal run:

```
uvicorn fastapi_app:app --reload --port 8000
```

Then open the Streamlit UI in another terminal:

```
streamlit run streamlit_app.py
```
